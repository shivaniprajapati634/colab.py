### Exercise 1: Creating Pandas Series

1. Create a Pandas Series containing the marks of 5 students in Maths.
2. Create a Pandas Series that stores 5 city names.
3. Create a Pandas Series with 5 different floating-point numbers representing temperature readings.

```python
#answer here
import pandas as pd

# 1. Pandas Series containing marks of 5 students in Maths
math_marks = pd.Series([85, 92, 78, 90, 88])
print("Maths Marks:\n", math_marks)

# 2. Pandas Series storing 5 city names
cities = pd.Series(["Delhi", "Mumbai", "Kolkata", "Chennai", "Bangalore"])
print("\nCities:\n", cities)

# 3. Pandas Series with 5 floating-point numbers representing temperature readings
temperatures = pd.Series([32.5, 28.4, 30.0, 33.2, 29.8])
print("\nTemperature Readings:\n", temperatures)
```

### Exercise 4: Saving DataFrames to CSV and Excel

1. Save any one of your created DataFrames into a CSV file named 'students.csv'.
2. Save another DataFrame into an Excel file named 'products.xlsx'.

```python
#answer here
import pandas as pd

# Assuming these are your DataFrames
# Students DataFrame
students_df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 22],
    "City": ["Delhi", "Mumbai", "Bangalore"]
})

# Products DataFrame
products_df = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet"],
    "Category": ["Electronics", "Electronics", "Electronics"],
    "Price": [50000, 15000, 20000]
})

# 1. Save students_df to CSV
students_df.to_csv("students.csv", index=False)
print("students.csv saved successfully!")

# 2. Save products_df to Excel
products_df.to_excel("products.xlsx", index=False)
print("products.xlsx saved successfully!")
```

### Exercise 5: Reading Data from Files
